```python
from hypertiling import HyperbolicTiling
import json

# Parameters
p = 4
q = 5
nlayers = 3

# Generate the tiling
T = HyperbolicTiling(p, q, nlayers)

# Access the full list of tiles
tiles_data = []
for tile in T.polygons:
    neighbors = T.get_nbrs(tile.idx)
    tile_data = {
        "id": int(tile.idx),  # Convert from np.int64 to Python int
        "neighbors": [int(n) if n is not None else None for n in neighbors],  # Same here
    }
    tiles_data.append(tile_data)

# Save to JSON
with open("tiling.json", "w") as f:
    json.dump(tiles_data, f, indent=2)
```

```python
def myplot(tiling):
    ax = plot_geodesic(T, ec="0.6", lw=0.7, zorder=3)

    # in the SRG kernel, the polygons are stored in a dictionary
    # hence we need to iterate over the keys which are the indices
    for idx in tiling.polygons.keys():
        if idx in T.exposed:
            col = "#ad153d"
        else:
            col = "k"

        # extract center coordinate and display index there
        z = tiling.get_center(idx)
        t = ax.text(np.real(z), np.imag(z), str(idx), c=col, ha="center", va="center", fontsize=17*(1-0.7*np.abs(z)), zorder=1)
        t.set_bbox(dict(facecolor='w', edgecolor="none", alpha=0.5))
```
